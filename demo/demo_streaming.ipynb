```python
from app.core.llm import get_llm, extract_text_content
from app.config import settings

print(f"Configured Provider: {settings.LLM_PROVIDER}")
print(f"Configured Model: {settings.LLM_MODEL}")

llm = get_llm()

print(f"llm initialized !")
```

```python
for chunk in llm.stream("What is the capital of India?"):
    text = extract_text_content(chunk.content)
    print("Chunk:", chunk)
    print("Type:", type(chunk))
    print("Raw Content:", repr(chunk.content))
    print("Extracted Text:", repr(text))
    print("-" * 50)
```
